## Authenticated live-data cell (research environment only)
This cell calls the planned trial-scoped analyst endpoint only when the private Jupyter environment provides short-lived credentials. It displays a count and does not print participant-level records. The route and authorization service are not deployed yet.


```python
import os
import requests

TRIAL_SLUG = "anxiety"
api_base = os.getenv("TRIALS_API_URL")
token = os.getenv("TRIALS_ANALYST_TOKEN")
if not api_base or not token:
    print("Live data is not configured. Synthetic example only.")
else:
    response = requests.get(
        f"{api_base.rstrip('/')}/api/analyst/trials/{TRIAL_SLUG}/responses/summary",
        headers={"Authorization": f"Bearer {token}"}, timeout=20
    )
    response.raise_for_status()
    summary = response.json()
    print({"trial": TRIAL_SLUG, "protocol_version": summary.get("protocol_version"),
           "snapshot_id": summary.get("snapshot_id"), "record_count": summary.get("record_count")})
```
